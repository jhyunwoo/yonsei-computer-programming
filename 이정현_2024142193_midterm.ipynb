## 1.1
코드 접근방법: input 함수로 기준 값을 입력 받고 while문으로 loop를 형성해 기준 값을 만족할 때 코드가 계속 반복되게 실행 될 수 있도록 설정했습니다. 각각 수를 세기 위한 값들을 초깃값 0과 함께 설정하고 입력받은 기준 값을 만족할 때 문자열을 입력 받았습니다. 문자열을 list로 받아들이면 이때 각 인덱스를 for 문을 적용해 양옆에 값을 비교했습니다. 

```python
## Your Code ##
num =int(input("문자열의 개수:")) #문자열의 개수, 코드의 반복 횟수를 입력받음.
trial = 0 #시도의 초깃값 0
count1 = 0 #YN의 개수의 초깃값 0 
count2 = 0 #NY의 개수의 초깃값 0


while trial < num: #반복해야 하는 횟수보다 시도의 횟수가 적으면 계속 되는 loop를 설정함. 
    string = input("문자열을 입력하세요:") #문자열을 list 형태로 받음.
    for i in range(len(string)-1): #for 문을 활용해 문자열  list의 길이 만큼의 range of numbers를 대표하는 i를 설정함.
        if string [i] == string [i+1]: #양옆 문자가 같으면 continue하도록 설정함.
            continue
        elif string [i] == "Y": #양옆문자가 같지 않고 첫 문자가 Y이면 count1에 1을 적립함.(update)
            if string[i+1] !=string[i]:
                count1+=1
            else: pass
        elif string [i] == "N": #양옆문자가 같지 않고 첫 문자가 N이면 count2에 1을 적립함.(update)
            if string[i+1] !=string[i]:
                count2+=1
            else: pass
    print(f"YN의 개수: {count1}, NY의 개수: {count2}") #while문이 마무리 되면서 결과를 출력할 수 있도록 f-string을 활용해 코딩함.
    trial+=1 #시도의 수를 세기 위한 업데이트 코드임.
```

## 1.2
코드 접근방법: 문자열을 input 함수로 입력받고 각 수열을 이루고 있는 수를 int format으로 변환한 list를 만들었습니다. 이때 이 리스트의 길이보다 시도 숫자가 작다는 것이 참이면 list 안에 있는 각 수를 뒤에 있는 수와 비교해 letters라는 list에 조건에 맞는 Y 또는 N을 추가했습니다. 완성된 letters list를 위 코드를 인용해 각각 YN NY의 개수를 세고, 결국 한 문자열로 정리한 후 결과를 출력했습니다.

```python
## Your Code ##
series = input("1부터 9까지의 값으로 이루어진 수열을 입력하세요(단 ','로 구분):").split(",") #수열을 각각 str으로 ","로 구분해 list로 받음.
for i in range(len(series)): #각 인덱스를 정수 format으로 변환함.
    series[i] = int(series[i])
trial =0 #시도의 초깃값 0을 설정함.

while trial < len(series): #시도 횟수가 series의 길이보다 작다는 점을 만족하면 반복되도록 코딩함.
    letters =[]
    for i in range (len(series) -1): #각 인덱스를 비교하기 위해 for 문을 도입함.
        if series[i]< series[i+1]: #뒤에 인덱스와 크기를 비교했을 때 Y 조건에 만족시 letters 리스트에 Y를 추가함.
            letters.append("Y")
        elif series[i] > series [i+1]:#뒤에 인덱스와 크기를 비교했을 때 N 조건에 만족시 letters 리스트에 N를 추가함.
            letters.append("N")
        else: pass
        trial+=1 #시도 마다 시도 횟수가 업데이트 되도록 함.

count1 = 0 #YN의 개수의 초깃값 0 
count2 = 0 #NY의 개수의 초깃값 0

for i in range(len(letters)-1): #for 문을 활용해 문자열  list의 길이 만큼의 range of numbers를 대표하는 i를 설정함.
        if letters [i] == letters [i+1]: #양옆 문자가 같으면 continue하도록 설정함.
            continue
        elif letters [i] == "Y": #양옆문자가 같지 않고 첫 문자가 Y이면 count1에 1을 적립함.(update)
            if letters[i+1] !=letters[i]:
                count1+=1
            else: pass
        elif letters [i] == "N": #양옆문자가 같지 않고 첫 문자가 N이면 count2에 1을 적립함.(update)
            if letters[i+1] !=letters[i]:
                count2+=1
            else: pass

for i in range(1,len(letters)): #letters의 인덱스들을 하나의 string으로 합치기 위한 for 문.
    letters[0] += letters[i]

print ("문자열 결과: {}".format(letters[0])) #앞서 합친 문자열을 출력함.
print(f"YN의 개수: {count1}, NY의 개수: {count2}") #while문이 마무리 되면서 결과를 출력할 수 있도록 f-string을 활용해 코딩함.
```

# 2.
코드 접근방법: 학생등의 초기 정보를 studentinfo dictionary에 저장하고 while문을 형성해 loop을 제작합니다. 명령어를 입력 받고if문을 활용해 명령문이 무엇이냐에 따라 각각 조건에 맞는 결과를 출력하도록 했습니다. 마지막 end에 break을 쓰면서 명령어 end 입력 시 loop이 깨지도록 설정했습니다.

```python
## Your Code ##
studentinfo ={ "김연세": 2024142001, "이전기":2024142002} #초기 학생 정보 dictionary로 정리함.

while True: #loop 형성함.
    C = input("명령어를 입력하세요(register, print, search, end):") #각 네가지 종류의 명령어 중 명령어를 받음.
    if C =="register": #register 일때 이름과 학번을 입력받을 수 있도록함.
        name = input("이름을 등록하세요:")
        id = int(input("학번을 등록하세요:"))
        studentinfo_2 ={name : id} #입력 받은 정보로 새로운 dictionary를 정의함.
        studentinfo.update(studentinfo_2) #새로운 dictionary를 기존에 있던 dictionary에 업데이트함.
        print("{}의 학번 {}가 저장되었습니다.".format(name, id)) #결과를 출력함.
        continue #loop을 이어나감.
    
    elif C == "print": #print 명령어를 받았을 때 dictionary를 출력함.
        print(studentinfo)
    elif C=="search": #search 명령어를 받았을 때 이름을 입력 받음.
        q=input("검색할 이름을 입력하세요:")
        if q in studentinfo: #이름이 studentinfo에 있으면 학번 정보를 출력함.
            o=studentinfo [q]
            print("{}의 학번은 {}입니다.".format(q, o))
        else:  #아니면 정보가 없다고 출력함.
            print(f"{q}의 정보가 없습니다.")
    elif C =="end": #end 명령어 일때 loop을 종료함.
        break
    else: #그외에 명령어 일때 다시 입력하라는 말을 출력함.
        print("잘못된 명령입니다. 다시 입력하세요.")
           
```

## 3.
코드 접근방법: 각각 Character class를 형성해 각 초깃값을 초기화 하고 damag_method를 설정했습니다. 이의 child class로 각각 공격력, 방어력, 직업을 설정해 주고 그대로 불러와 저장했습니다. 그리고 battle 함수를 정의해 두 character가 싸우는 과정을 정의했습니다.

```python
## Your Code ##
class Character:
    job =""
    health =100
    def __init__(self, job,health):
        self.job =job
        self.health =health

    def init_method(self, atk, dfs):
        self.atk =atk
        self.dfs =dfs

    def damage_method (self, atk_opponent):
        self.atk_opponent = atk_opponent
        health = self.health - (int(atk_opponent))*(100 -int(self.dfs))/100
        if health>0:
            print("Warrior (or Wizard)의 남은 체력: {}".format(health))
        else:
            print("Warrior (or Wizard) 패배")

class Warrior(Character):
    atk = 50
    dfs = 60
    job ="Warrior"
    def __init__(self, atk, dfs, job):
        super().init_method(atk, dfs)
        super().__init__(job) 

    def damage_method (self, atk_opponent):
        super().damage_method(atk_opponent)


class Wizard (Character):
    atk = 70
    dfs = 40
    job ="Wizard"
    def __init__(self, atk, dfs, job):
        super().init_method(atk, dfs)
        super().__init__(job)
    def damage_method (self, atk_opponent):
        super().damage_method(atk_opponent)


def battle(Warrior, Wizard):
    turn = 1
    while any Character.health !=0:
        if turn % 2!=0:
            Warrior.damage_method(atk)
            turn+=1
        else:
            Wizard.damage_method(atk)
            turn+=1
```
